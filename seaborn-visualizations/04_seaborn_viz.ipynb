# Visualisation avancée avec Seaborn 📊
Créer des graphiques informatifs et esthétiques
sur le dataset Titanic.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Style global
sns.set_theme(style="whitegrid", palette="muted")

# Charger le dataset
df = pd.read_csv(r"C:\Users\Eartrh\Documents\Etudes_France_2025\Science_De_Données\data-analysis-projects\titanic-eda\data\titanic.csv")
print(f"✅ Dataset chargé — {df.shape[0]} passagers")
```

```python
sns.countplot(data=df, x='Sex', hue='Survived')
plt.title("Survie par sexe")
plt.legend(['Décédé', 'Survivant'])
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=df, x='Pclass', y='Age', palette='muted')
plt.title("Distribution des âges par classe")
plt.xlabel("Classe")
plt.ylabel("Age")
plt.tight_layout()
plt.show()
```

```python
# Heatmap des corrélations entre variables numériques
corr = df[['Survived','Pclass','Age','SibSp','Parch','Fare']].corr()

sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', center=0)
plt.title("Matrice de corrélation")
plt.tight_layout()
plt.show()
```

```python
# Nombre de survivants par sexe
plt.figure(figsize=(7,4))
sns.countplot(data=df, x='Sex', hue='Survived',
              palette=['#E74C3C','#2ECC71'])
plt.title("Survivants par sexe")
plt.legend(['Décédé','Survivant'])
plt.tight_layout()
plt.show()
```

```python
# Distribution des âges par classe
plt.figure(figsize=(7,4))
sns.boxplot(data=df, x='Pclass', y='Age',
            palette='muted')
plt.title("Distribution des âges par classe")
plt.xlabel("Classe")
plt.ylabel("Age")
plt.tight_layout()
plt.show()
```

```python
# Corrélations entre variables numériques
plt.figure(figsize=(7,5))
sns.heatmap(df[['Survived','Pclass','Age','SibSp','Fare']].corr(),
            annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title("Matrice de corrélation")
plt.tight_layout()
plt.show()
```

```python
# Distribution de l'âge par classe
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x='Pclass', y='Age', palette='muted')
plt.title("Distribution de l'âge par classe")
plt.xlabel("Classe")
plt.ylabel("Age")
plt.xticks([0,1,2], ['1ère classe','2ème classe','3ème classe'])
plt.show()
```

```python
# Survie par sexe et classe combinées
plt.figure(figsize=(8,5))
sns.violinplot(data=df, x='Pclass', y='Age', 
               hue='Survived', split=True, palette='coolwarm')
plt.title("Distribution âge/classe selon la survie")
plt.xlabel("Classe")
plt.ylabel("Age")
plt.xticks([0,1,2], ['1ère classe','2ème classe','3ème classe'])
plt.legend(['Décédé','Survivant'])
plt.show()
```

```python
# Nombre de survivants par sexe et classe
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Pclass', hue='Survived', 
              palette=['#E74C3C','#2ECC71'])
plt.title("Survivants vs Décédés par classe")
plt.xlabel("Classe")
plt.xticks([0,1,2], ['1ère classe','2ème classe','3ème classe'])
plt.legend(['Décédé','Survivant'])
plt.show()
```
