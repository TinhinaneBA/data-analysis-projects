# Nettoyage de données avec Pandas 🧹
Compétence clé du Data Analyst : transformer des données
"sales" en données exploitables.

````python
import pandas as pd
import numpy as np

# Création d'un dataset volontairement "sale"
data = {
    'nom':      ['Alice', 'Bob', 'Alice', 'David', None, 'Frank', 'Bob', 'Helen'],
    'age':      [25, 'trente', 22, 45, 31, -5, 28, 200],
    'ville':    ['Paris', 'paris', 'Paris', 'LYON', 'Marseille', 'Lyon', 'paris', 'Bordeaux'],
    'salaire':  [3000, 2500, 3000, None, 4000, 2800, 2500, 3500],
    'email':    ['alice@mail.com', 'bob@mail.com', 'alice@mail.com', 
                 'david@mail.com', 'eve@mail.com', 'not-an-email', 
                 'bob@mail.com', 'helen@mail.com']
}

df = pd.DataFrame(data)
print("Dataset brut :")
df
````

````python
print(f"Lignes avant : {len(df)}")
df = df.drop_duplicates()
print(f"Lignes après suppression des doublons : {len(df)}")
````

````python
print("Valeurs manquantes :\n")
print(df.isnull().sum())

# Supprimer les lignes sans nom
df = df.dropna(subset=['nom'])

# Remplir les salaires manquants par la médiane
df['salaire'] = df['salaire'].fillna(df['salaire'].median())

print("\nAprès nettoyage :")
print(df.isnull().sum())
````

````python
# Convertir en numérique — les textes deviennent NaN
df['age'] = pd.to_numeric(df['age'], errors='coerce')

# Supprimer les âges impossibles
df = df[df['age'].between(0, 120)]

print(df['age'])
````

````python
# Garder uniquement les emails avec @ et .
mask = df['email'].str.contains('@') & df['email'].str.contains('\.')
df = df[mask]
print(df['email'])
````

````python
print("✅ Dataset nettoyé :")
print(f"Lignes : {len(df)}")
df
````

````python
# Comparaison avant/après
print("=" * 40)
print("RÉCAPITULATIF DU NETTOYAGE")
print("=" * 40)
print(f"Lignes initiales        : 8")
print(f"Lignes après nettoyage  : {len(df)}")
print(f"Supprimées              : {8 - len(df)}")
print()
print("Opérations effectuées :")
print("  ✅ Doublons supprimés")
print("  ✅ Valeurs manquantes traitées")
print("  ✅ Ages impossibles filtrés")
print("  ✅ Villes normalisées")
print("  ✅ Emails invalides supprimés")
print()
print("Dataset final :")
print(df.dtypes)
````

## Ce que j'ai appris 🧹

### Techniques de nettoyage appliquées

| Problème | Solution Pandas |
|---|---|
| Doublons | `drop_duplicates()` |
| Valeurs manquantes | `dropna()` / `fillna()` |
| Type incorrect | `pd.to_numeric(errors='coerce')` |
| Valeurs impossibles | `df[df['col'].between(a, b)]` |
| Casse incohérente | `str.capitalize()` |
| Format invalide | `str.contains()` |

### Règle d'or
Dans la vraie vie, 80% du travail d'un Data Analyst
c'est le nettoyage. Des données propres = des analyses fiables.
